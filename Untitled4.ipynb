```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score
from sklearn.decomposition import PCA
```

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/x2ysrhky4j-lgtm/rasha-googlecolap-project/main/diabetes_prediction_dataset.csv'

df = pd.read_csv(url)

print("Dataset loaded successfully from GitHub!")
```

```python
print("Missing Values:\n", df.isnull().sum())
print("\nTotal Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print(f"Shape after removing duplicates: {df.shape}")
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(x='diabetes', data=df, palette='magma')
plt.title('Diabetes Status (0: No, 1: Yes)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='age', y='blood_glucose_level', hue='diabetes', data=df, alpha=0.5)
plt.title('Age vs Glucose Level')
plt.show()
```

```python
le = LabelEncoder()
df['gender'] = le.fit_transform(df['gender'])
df['smoking_history'] = le.fit_transform(df['smoking_history'])
print("Text data converted to numbers.")
```

```python
X = df.drop('diabetes', axis=1)
y = df['diabetes']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lr_model = LogisticRegression()
lr_model.fit(X_train_scaled, y_train)
print("Logistic Regression Accuracy:", accuracy_score(y_test, lr_model.predict(X_test_scaled)))
```

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_scaled, y_train)
print("Random Forest Accuracy:", accuracy_score(y_test, rf_model.predict(X_test_scaled)))
```

```python
from sklearn.svm import LinearSVC

# سنأخذ عينة من 5000 صف فقط للتدريب لجعل التنفيذ سريعاً جداً
X_train_sub, _, y_train_sub, _ = train_test_split(X_train_scaled, y_train, train_size=5000, stratify=y_train, random_state=42)

svm_model = SVC(kernel='rbf', C=1.0)
svm_model.fit(X_train_sub, y_train_sub)
print("Done!")
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(confusion_matrix(y_test, svm_model.predict(X_test_scaled)), annot=True, fmt='d', cmap='Greens')
plt.title('Confusion Matrix - SVM Model')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

```python
pca = PCA(n_components=2)
# رفعنا العينة هنا لتعطي نتائج أدق في الرسم والتحليل
X_train_pca = pca.fit_transform(X_train_scaled[:10000])
X_test_pca = pca.transform(X_test_scaled[:2000])
print("Dimensionality Reduction Complete with 10,000 samples!")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# 1. تجهيز الموديلات في قائمة
classifiers = {
    "SVM (Our Model)": svm_model, # الموديل الذي دربناه سابقاً
    "Logistic Regression": LogisticRegression(max_iter=1000, class_weight='balanced'),
    "Random Forest": RandomForestClassifier(n_estimators=50, class_weight='balanced')
}

print("--- Accuracy Comparison Report ---")

# 2. تدريب الموديلات الجديدة وقياس الدقة
for name, clf in classifiers.items():
    if name != "SVM (Our Model)": # لكي لا نعيد تدريب SVM ونضيع وقتنا
        clf.fit(X_train_scaled[:5000], y_train[:5000])

    # قياس الدقة على بيانات الاختبار
    score = clf.score(X_test_scaled[:1000], y_test[:1000])
    print(f"{name}: {score:.2%}")

print("---------------------------------")
```

```python
final_svm = SVC(kernel='rbf', C=1.0, class_weight='balanced')
final_svm.fit(X_train_pca, y_train[:10000])

y_pred_pca = final_svm.predict(X_test_pca)
print("Final Accuracy (Balanced Model):", accuracy_score(y_test[:2000], y_pred_pca))
```

```python

samples = np.array([
    [1, 65, 1, 1, 3, 35.0, 9.0, 200], # حالة مريض مؤكدة (سكر 200 وتراكمي 9)
    [0, 20, 0, 0, 4, 21.0, 4.0, 85],  # حالة سليم مؤكدة
    [1, 55, 0, 0, 2, 32.0, 8.5, 180]  # حالة مريض أخرى
])

samples_scaled = scaler.transform(samples)
samples_pca = pca.transform(samples_scaled)
predictions = final_svm.predict(samples_pca)

for i, pred in enumerate(predictions):
    status = "Diabetes Detected" if pred == 1 else "No Diabetes"
    print(f"Sample {i+1} Result: {status}")
```
